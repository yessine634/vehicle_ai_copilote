```python
import requests

url = "https://valhalla1.openstreetmap.de/route"

payload = {
    "locations": [
        {
            "lat": 34.7406,
            "lon": 10.7603
        },
        {
            "lat": 36.8065,
            "lon": 10.1815
        }
    ],
    "costing": "auto",
    "units": "kilometers"
}

headers = {
    "Content-Type": "application/json",
    "X-Client-Id": "vehicle-ai-copilot"
}

response = requests.post(
    url,
    json=payload,
    headers=headers,
    timeout=20
)

print("Status:", response.status_code)

response.raise_for_status()

data = response.json()

print(data)
```

```python
from pathlib import Path
from pprint import pprint
import sys

# Make the project package importable whether Jupyter starts in the
# project root or in the tests directory.
project_root = Path.cwd()
if not (project_root / "tools").is_dir():
    project_root = project_root.parent

if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from tools.routing import get_route

origin = "Sfax"
destination = "Tunis"

try:
    route = get_route(origin, destination)

    print(f"Route: {route['origin']['name']} -> {route['destination']['name']}")
    print(f"Distance: {route['distance_km']} km")
    print(f"Estimated duration: {route['duration_minutes']} minutes")
    print(f"Uses highway: {route['uses_highway']}")
    print(f"Has toll: {route['has_toll']}")
    print(f"Has ferry: {route['has_ferry']}")
    print("\nComplete result:")
    pprint(route, sort_dicts=False)
except (RuntimeError, ValueError) as error:
    print(f"Routing test failed: {error}")
```
